```python
"""
Breast Cancer Wisconsin Dataset - Comprehensive Classification Analysis
Paper-Ready Implementation with Statistical Analysis and Nature Medicine Standards

KAGGLE READY! 🚀
- Optimized for Kaggle environment
- Uses sklearn's built-in dataset (no upload needed)
- Just copy and run!

To use in Kaggle:
1. Create a new notebook
2. Copy this entire code
3. Run it!
"""

import pandas as pd
import numpy as np
import warnings
warnings.filterwarnings('ignore')

from sklearn.model_selection import KFold, cross_val_score, cross_val_predict, train_test_split
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier, GradientBoostingClassifier
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import (accuracy_score, precision_score, recall_score, f1_score, 
                            roc_auc_score, confusion_matrix, classification_report, roc_curve)

import xgboost as xgb
import lightgbm as lgb
import catboost as cb

from scipy import stats
from scipy.stats import wilcoxon, friedmanchisquare

import matplotlib.pyplot as plt
import seaborn as sns
from datetime import datetime
import os

try:
    get_ipython()
    IN_NOTEBOOK = True
    get_ipython().run_line_magic('matplotlib', 'inline')
except:
    IN_NOTEBOOK = False

# ==================== 1. LOAD DATA FROM KAGGLE ====================
print("="*110)
print("BREAST CANCER CLASSIFICATION - COMPREHENSIVE ANALYSIS")
print("="*110)
print("\n✓ Running on Kaggle environment")
print("-"*110)

# Load data from Kaggle directory
print("\nLoading Breast Cancer Dataset from Kaggle directory...")

# Try to load from Kaggle input directory
kaggle_paths = [
    "/kaggle/input/breast-cancer-wisconsin-original/wdbc.data",
    "/kaggle/input/breast-cancer-wisconsin-original/breast-cancer-wisconsin.data",
    "/kaggle/input/breast-cancer-wisconsin-original/wpbc.data"
]

data_loaded = False
dataset_type = None

for path in kaggle_paths:
    if os.path.exists(path):
        print(f"✓ Found dataset at: {path}")
        
        if 'wdbc.data' in path:
            # WDBC format (Diagnostic) - 30 features
            print("  Loading WDBC (Diagnostic) format...")
            column_names = ['ID', 'Diagnosis']
            feature_types = ['mean', 'se', 'worst']
            measurements = ['radius', 'texture', 'perimeter', 'area', 'smoothness', 
                          'compactness', 'concavity', 'concave_points', 'symmetry', 'fractal_dimension']
            for measure in measurements:
                for ftype in feature_types:
                    column_names.append(f'{measure}_{ftype}')
            df = pd.read_csv(path, header=None, names=column_names)
            dataset_type = 'wdbc'
            
        elif 'breast-cancer-wisconsin.data' in path:
            # Original format - 9 features
            print("  Loading Original Wisconsin format...")
            column_names = ['ID', 'Clump_Thickness', 'Uniformity_Cell_Size', 'Uniformity_Cell_Shape',
                           'Marginal_Adhesion', 'Single_Epithelial_Cell_Size', 'Bare_Nuclei', 
                           'Bland_Chromatin', 'Normal_Nucleoli', 'Mitoses', 'Class']
            df = pd.read_csv(path, header=None, names=column_names)
            # Handle missing values (marked as '?')
            df = df.replace('?', np.nan)
            df = df.dropna()
            # Rename Class to Diagnosis for consistency
            df = df.rename(columns={'Class': 'Diagnosis'})
            dataset_type = 'original'
            
        elif 'wpbc.data' in path:
            # WPBC format (Prognostic)
            print("  Loading WPBC (Prognostic) format...")
            wpbc_columns = ['ID', 'Outcome', 'Time']
            feature_types = ['mean', 'se', 'worst']
            measurements = ['radius', 'texture', 'perimeter', 'area', 'smoothness', 
                           'compactness', 'concavity', 'concave_points', 'symmetry', 'fractal_dimension']
            for measure in measurements:
                for ftype in feature_types:
                    wpbc_columns.append(f'{measure}_{ftype}')
            wpbc_columns.extend(['Tumor_Size', 'Lymph_Node_Status'])
            df = pd.read_csv(path, header=None, names=wpbc_columns)
            # Handle missing values
            df = df.replace('?', np.nan)
            df = df.dropna()
            # Rename Outcome to Diagnosis for consistency
            df = df.rename(columns={'Outcome': 'Diagnosis'})
            # Drop Time column
            if 'Time' in df.columns:
                df = df.drop('Time', axis=1)
            dataset_type = 'wpbc'
        
        data_loaded = True
        break

# If no Kaggle dataset found, use sklearn built-in as fallback
if not data_loaded:
    print("⚠ No dataset found in Kaggle directory. Using sklearn built-in dataset...")
    from sklearn.datasets import load_breast_cancer
    data = load_breast_cancer()
    df = pd.DataFrame(data.data, columns=data.feature_names)
    df['Diagnosis'] = data.target
    df['Diagnosis'] = df['Diagnosis'].map({0: 'M', 1: 'B'})  # 0=malignant, 1=benign in sklearn
    dataset_type = 'sklearn'

print(f"✓ Dataset loaded successfully!")
print(f"  Shape: {df.shape}")
print(f"  Columns: {df.columns.tolist()[:5]}... (showing first 5)")

# ==================== 2. DATA EXPLORATION AND CLEANING ====================
print("\n" + "="*110)
print("DATA EXPLORATION AND CLEANING")
print("="*110)

# Display basic info
print("\nDataset Info:")
print(f"  Total Rows: {df.shape[0]}")
print(f"  Total Columns: {df.shape[1]}")
print(f"\nFirst few rows:")
print(df.head())

print(f"\nData Types:")
print(df.dtypes)

print(f"\nMissing Values:")
missing = df.isnull().sum()
if missing.sum() > 0:
    print(missing[missing > 0])
else:
    print("  No missing values found!")

# Identify target column (diagnosis)
target_col = None
for col in ['Diagnosis', 'diagnosis', 'target', 'class', 'Class']:
    if col in df.columns:
        target_col = col
        break

if target_col is None:
    print("\n⚠ Warning: Could not find diagnosis column. Please check your dataset.")
    print(f"Available columns: {df.columns.tolist()}")
    raise ValueError("Diagnosis column not found")

print(f"\nTarget Column: '{target_col}'")
print(f"Unique values in target: {df[target_col].unique()}")
print(f"Class Distribution:")
print(df[target_col].value_counts())

# Clean and prepare data
# Remove ID column if exists
id_cols = ['id', 'ID', 'Id']
for col in id_cols:
    if col in df.columns:
        df = df.drop(col, axis=1)
        print(f"✓ Removed ID column: {col}")

# Remove unnamed columns
df = df.loc[:, ~df.columns.str.contains('^Unnamed')]

# Convert diagnosis to binary based on dataset type
print(f"\nConverting diagnosis to binary format...")
if dataset_type == 'original':
    # Original: 2=Benign, 4=Malignant
    df[target_col] = df[target_col].astype(int)
    df[target_col] = df[target_col].map({2: 0, 4: 1})  # 0=Benign, 1=Malignant
    class_names = ['Benign', 'Malignant']
    print(f"  Mapped: 2→Benign(0), 4→Malignant(1)")
elif dataset_type == 'wpbc':
    # WPBC: N=No Recurrence, R=Recurrence
    le = LabelEncoder()
    df[target_col] = le.fit_transform(df[target_col])
    class_names = ['No Recurrence', 'Recurrence'] if le.classes_[0] in ['N', 'n'] else ['Recurrence', 'No Recurrence']
    print(f"  Encoded: {le.classes_[0]}→{class_names[0]}(0), {le.classes_[1]}→{class_names[1]}(1)")
else:
    # WDBC or sklearn: M=Malignant, B=Benign
    le = LabelEncoder()
    df[target_col] = le.fit_transform(df[target_col])
    # Check the encoding
    if le.classes_[0] in ['B', 'b', 'benign', 0]:
        class_names = ['Benign', 'Malignant']
    else:
        class_names = ['Malignant', 'Benign']
    print(f"  Encoded: {le.classes_[0]}→{class_names[0]}(0), {le.classes_[1]}→{class_names[1]}(1)")

# Verify we have both classes
unique_classes = df[target_col].unique()
if len(unique_classes) < 2:
    print(f"\n⚠ ERROR: Only {len(unique_classes)} class found in data!")
    print(f"  Unique values: {unique_classes}")
    raise ValueError("Dataset must have at least 2 classes for classification")

# Separate features and target
X = df.drop(target_col, axis=1).values
y = df[target_col].values
feature_names = df.drop(target_col, axis=1).columns.tolist()

print(f"\n✓ Data prepared successfully!")
print(f"  Samples: {X.shape[0]}")
print(f"  Features: {X.shape[1]}")
print(f"  Classes: {class_names[0]} ({np.sum(y==0)}), {class_names[1]} ({np.sum(y==1)})")
print(f"  Class distribution: {np.sum(y==0)/len(y)*100:.1f}% / {np.sum(y==1)/len(y)*100:.1f}%")

# ==================== 3. DATA VISUALIZATION ====================
print("\n" + "="*110)
print("DATA VISUALIZATION")
print("="*110)

# Create output directory
output_dir = "breast_cancer_classification"
os.makedirs(output_dir, exist_ok=True)
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

# Set Nature Medicine standards for visualizations
plt.rcParams.update({
    'font.family': 'sans-serif',
    'font.sans-serif': ['Arial', 'Helvetica', 'DejaVu Sans', 'Liberation Sans', 'sans-serif'],
    'font.size': 7,
    'axes.labelsize': 7,
    'axes.titlesize': 8,
    'xtick.labelsize': 6,
    'ytick.labelsize': 6,
    'legend.fontsize': 6,
    'figure.dpi': 300,
    'savefig.dpi': 300,
    'axes.linewidth': 0.5,
    'xtick.major.width': 0.5,
    'ytick.major.width': 0.5,
})

DOUBLE_COLUMN_WIDTH = 7.2
NATURE_COLORS = ['#0173B2', '#DE8F05', '#029E73', '#CC78BC', '#CA9161', '#949494', '#ECE133', '#56B4E9']
sns.set_palette(NATURE_COLORS)

# Visualization 1: Class Distribution
print("\n[1/5] Creating class distribution plot...")
fig, axes = plt.subplots(1, 2, figsize=(DOUBLE_COLUMN_WIDTH * 0.6, DOUBLE_COLUMN_WIDTH * 0.3))

# Count plot
class_counts = pd.Series(y).value_counts().sort_index()
n_classes = len(class_counts)

# Ensure we have the right number of class names
if n_classes != len(class_names):
    print(f"⚠ Warning: Mismatch between classes ({n_classes}) and class names ({len(class_names)})")
    class_names = [f'Class {i}' for i in range(n_classes)]

class_labels = [class_names[i] for i in range(n_classes)]
axes[0].bar(class_labels, class_counts.values, 
           color=NATURE_COLORS[:n_classes], alpha=0.85, edgecolor='black', linewidth=0.5)
axes[0].set_ylabel('Count', fontsize=7)
axes[0].set_title('Class Distribution', fontsize=8, fontweight='bold')
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)
axes[0].grid(axis='y', linestyle='--', alpha=0.3, linewidth=0.5)
for i, v in enumerate(class_counts.values):
    axes[0].text(i, v + 5, str(v), ha='center', va='bottom', fontsize=6, fontweight='bold')

# Pie chart
pie_labels = [f'{class_names[i]}\n({class_counts.values[i]})' for i in range(n_classes)]
axes[1].pie(class_counts.values, labels=pie_labels,
           colors=NATURE_COLORS[:n_classes], autopct='%1.1f%%',
           startangle=90, textprops={'fontsize': 6})
axes[1].set_title('Class Proportion', fontsize=8, fontweight='bold')

plt.tight_layout()
plt.savefig(f"{output_dir}/DataViz_01_Class_Distribution_{timestamp}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{output_dir}/DataViz_01_Class_Distribution_{timestamp}.eps", format='eps', bbox_inches='tight')
if IN_NOTEBOOK:
    plt.show()
else:
    plt.close()
print("✓ Class distribution plot saved")

# Visualization 2: Feature Statistics Summary
print("[2/5] Creating feature statistics summary...")
fig, ax = plt.subplots(figsize=(DOUBLE_COLUMN_WIDTH, DOUBLE_COLUMN_WIDTH * 0.4))

# Calculate statistics for each feature
feature_stats = pd.DataFrame({
    'Feature': feature_names[:min(15, len(feature_names))],
    'Mean': X[:, :min(15, len(feature_names))].mean(axis=0),
    'Std': X[:, :min(15, len(feature_names))].std(axis=0)
})

x_pos = np.arange(len(feature_stats))
ax.bar(x_pos, feature_stats['Mean'], yerr=feature_stats['Std'], 
       color=NATURE_COLORS[2], alpha=0.7, edgecolor='black', linewidth=0.5, capsize=3)
ax.set_xticks(x_pos)
ax.set_xticklabels(feature_stats['Feature'], rotation=45, ha='right', fontsize=5)
ax.set_ylabel('Value (Mean ± Std)', fontsize=7)
ax.set_title(f'Feature Statistics (First {len(feature_stats)} Features)', fontsize=8, fontweight='bold')
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='y', linestyle='--', alpha=0.3, linewidth=0.5)

plt.tight_layout()
plt.savefig(f"{output_dir}/DataViz_02_Feature_Statistics_{timestamp}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{output_dir}/DataViz_02_Feature_Statistics_{timestamp}.eps", format='eps', bbox_inches='tight')
if IN_NOTEBOOK:
    plt.show()
else:
    plt.close()
print("✓ Feature statistics plot saved")

# Visualization 3: Feature Distribution by Class (Box plots for top 6 features)
print("[3/5] Creating feature distribution by class...")
fig, axes = plt.subplots(2, 3, figsize=(DOUBLE_COLUMN_WIDTH, DOUBLE_COLUMN_WIDTH * 0.5))
axes = axes.flatten()

# Select top 6 features based on variance
feature_var = X.var(axis=0)
top_features_idx = np.argsort(feature_var)[-6:][::-1]

for idx, feat_idx in enumerate(top_features_idx):
    df_plot = pd.DataFrame({
        'Value': X[:, feat_idx],
        'Class': [class_names[int(i)] for i in y]
    })
    
    bp = axes[idx].boxplot([df_plot[df_plot['Class']==class_names[0]]['Value'],
                            df_plot[df_plot['Class']==class_names[1]]['Value']],
                           labels=[class_names[0], class_names[1]],
                           patch_artist=True, widths=0.6,
                           boxprops=dict(linewidth=0.5),
                           whiskerprops=dict(linewidth=0.5),
                           capprops=dict(linewidth=0.5),
                           medianprops=dict(linewidth=1.0, color='darkred'))
    
    for patch, color in zip(bp['boxes'], [NATURE_COLORS[0], NATURE_COLORS[1]]):
        patch.set_facecolor(color)
        patch.set_alpha(0.7)
    
    axes[idx].set_title(feature_names[feat_idx], fontsize=7, fontweight='bold')
    axes[idx].tick_params(labelsize=6)
    axes[idx].spines['top'].set_visible(False)
    axes[idx].spines['right'].set_visible(False)
    axes[idx].grid(axis='y', linestyle='--', alpha=0.3, linewidth=0.5)

plt.suptitle('Feature Distribution by Class (Top 6 Features)', fontsize=8, fontweight='bold', y=1.00)
plt.tight_layout()
plt.savefig(f"{output_dir}/DataViz_03_Feature_Distribution_{timestamp}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{output_dir}/DataViz_03_Feature_Distribution_{timestamp}.eps", format='eps', bbox_inches='tight')
if IN_NOTEBOOK:
    plt.show()
else:
    plt.close()
print("✓ Feature distribution plot saved")

# Visualization 4: Correlation Heatmap (for first 15 features)
print("[4/5] Creating correlation heatmap...")
n_features_to_show = min(15, len(feature_names))
fig, ax = plt.subplots(figsize=(DOUBLE_COLUMN_WIDTH * 0.7, DOUBLE_COLUMN_WIDTH * 0.6))

corr_matrix = np.corrcoef(X[:, :n_features_to_show].T)
im = ax.imshow(corr_matrix, cmap='coolwarm', aspect='auto', vmin=-1, vmax=1)

ax.set_xticks(np.arange(n_features_to_show))
ax.set_yticks(np.arange(n_features_to_show))
ax.set_xticklabels(feature_names[:n_features_to_show], rotation=45, ha='right', fontsize=5)
ax.set_yticklabels(feature_names[:n_features_to_show], fontsize=5)

cbar = plt.colorbar(im, ax=ax, shrink=0.8)
cbar.set_label('Correlation', fontsize=6)
cbar.ax.tick_params(labelsize=5)

ax.set_title(f'Feature Correlation Matrix (First {n_features_to_show} Features)', fontsize=8, fontweight='bold')

plt.tight_layout()
plt.savefig(f"{output_dir}/DataViz_04_Correlation_Heatmap_{timestamp}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{output_dir}/DataViz_04_Correlation_Heatmap_{timestamp}.eps", format='eps', bbox_inches='tight')
if IN_NOTEBOOK:
    plt.show()
else:
    plt.close()
print("✓ Correlation heatmap saved")

# Visualization 5: Data Quality Check
print("[5/5] Creating data quality summary...")
fig, axes = plt.subplots(1, 2, figsize=(DOUBLE_COLUMN_WIDTH * 0.6, DOUBLE_COLUMN_WIDTH * 0.3))

# Missing values
missing_counts = pd.Series(df.isnull().sum().sum(), index=['Total Missing'])
axes[0].bar(['Missing Values'], [missing_counts.values[0]], 
           color=NATURE_COLORS[3] if missing_counts.values[0] > 0 else NATURE_COLORS[2], 
           alpha=0.85, edgecolor='black', linewidth=0.5)
axes[0].set_ylabel('Count', fontsize=7)
axes[0].set_title('Data Completeness', fontsize=8, fontweight='bold')
axes[0].text(0, missing_counts.values[0] + 0.5, f'{int(missing_counts.values[0])}', 
            ha='center', va='bottom', fontsize=7, fontweight='bold')
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)
axes[0].set_ylim([0, max(10, missing_counts.values[0] * 1.2)])

# Feature count
axes[1].bar(['Features', 'Samples'], [X.shape[1], X.shape[0]], 
           color=[NATURE_COLORS[4], NATURE_COLORS[5]], alpha=0.85, edgecolor='black', linewidth=0.5)
axes[1].set_ylabel('Count', fontsize=7)
axes[1].set_title('Dataset Dimensions', fontsize=8, fontweight='bold')
axes[1].text(0, X.shape[1] + 5, str(X.shape[1]), ha='center', va='bottom', fontsize=7, fontweight='bold')
axes[1].text(1, X.shape[0] + 5, str(X.shape[0]), ha='center', va='bottom', fontsize=7, fontweight='bold')
axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)

plt.tight_layout()
plt.savefig(f"{output_dir}/DataViz_05_Data_Quality_{timestamp}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{output_dir}/DataViz_05_Data_Quality_{timestamp}.eps", format='eps', bbox_inches='tight')
if IN_NOTEBOOK:
    plt.show()
else:
    plt.close()
print("✓ Data quality summary saved")

print("\n✓ All data visualizations completed!")
print(f"✓ 5 visualization figures saved to: {output_dir}/")

# ==================== 4. SCALE DATA ====================
print("\n" + "="*110)
print("DATA PREPROCESSING")
print("="*110)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
print("✓ Data standardized using StandardScaler")
print(f"  Original data range: [{X.min():.2f}, {X.max():.2f}]")
print(f"  Scaled data range: [{X_scaled.min():.2f}, {X_scaled.max():.2f}]")

# ==================== 5. DEFINE BEST MODELS (OPTIMIZED) ====================
print("\n" + "="*110)
print("BEST CLASSIFICATION MODELS (Optimized Hyperparameters)")
print("="*110)

# Best performing models with tuned hyperparameters for classification
models = {
    # ===== GRADIENT BOOSTING MODELS (Best for tabular data) =====
    'XGBoost (Optimized)'       : xgb.XGBClassifier(
        n_estimators=500,
        max_depth=6,
        learning_rate=0.01,
        subsample=0.8,
        colsample_bytree=0.8,
        gamma=0.1,
        reg_alpha=0.1,
        reg_lambda=1.0,
        min_child_weight=3,
        random_state=42,
        n_jobs=-1,
        eval_metric='logloss'
    ),
    
    'LightGBM (Optimized)'      : lgb.LGBMClassifier(
        n_estimators=500,
        max_depth=6,
        learning_rate=0.01,
        subsample=0.8,
        colsample_bytree=0.8,
        num_leaves=31,
        min_child_samples=20,
        reg_alpha=0.1,
        reg_lambda=1.0,
        random_state=42,
        verbose=-1,
        n_jobs=-1
    ),
    
    'CatBoost (Optimized)'      : cb.CatBoostClassifier(
        iterations=500,
        depth=6,
        learning_rate=0.01,
        l2_leaf_reg=3,
        bootstrap_type='Bayesian',
        bagging_temperature=1,
        random_strength=1,
        verbose=False,
        random_state=42
    ),
    
    # ===== ENSEMBLE MODELS =====
    'Random Forest (Optimized)' : RandomForestClassifier(
        n_estimators=500,
        max_depth=10,
        min_samples_split=5,
        min_samples_leaf=2,
        max_features='sqrt',
        bootstrap=True,
        oob_score=True,
        random_state=42,
        n_jobs=-1
    ),
    
    'Extra Trees (Optimized)'   : ExtraTreesClassifier(
        n_estimators=500,
        max_depth=10,
        min_samples_split=5,
        min_samples_leaf=2,
        max_features='sqrt',
        bootstrap=True,
        random_state=42,
        n_jobs=-1
    ),
    
    'Gradient Boosting (Optimized)' : GradientBoostingClassifier(
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.8,
        min_samples_split=5,
        min_samples_leaf=2,
        max_features='sqrt',
        random_state=42
    ),
    
    # ===== SUPPORT VECTOR MACHINES =====
    'SVM (RBF Optimized)'       : SVC(
        kernel='rbf',
        C=10,
        gamma='scale',
        probability=True,
        class_weight='balanced',
        random_state=42
    ),
    
    'SVM (Linear Optimized)'    : SVC(
        kernel='linear',
        C=1.0,
        probability=True,
        class_weight='balanced',
        random_state=42
    ),
    
    # ===== LOGISTIC REGRESSION =====
    'Logistic Regression (L2)'  : LogisticRegression(
        penalty='l2',
        C=1.0,
        solver='lbfgs',
        max_iter=2000,
        class_weight='balanced',
        random_state=42,
        n_jobs=-1
    ),
    
    'Logistic Regression (L1)'  : LogisticRegression(
        penalty='l1',
        C=1.0,
        solver='liblinear',
        max_iter=2000,
        class_weight='balanced',
        random_state=42
    ),
    
    # ===== NEURAL NETWORK =====
    'MLP Neural Network (Deep)' : MLPClassifier(
        hidden_layer_sizes=(256, 128, 64, 32),
        activation='relu',
        solver='adam',
        alpha=0.0001,
        batch_size=32,
        learning_rate='adaptive',
        learning_rate_init=0.001,
        max_iter=2000,
        early_stopping=True,
        validation_fraction=0.1,
        n_iter_no_change=20,
        random_state=42
    ),
}

print(f"\n✓ {len(models)} optimized classification models ready")
print(f"✓ Models selected: State-of-the-art algorithms with tuned hyperparameters")

# ==================== 6. CROSS-VALIDATION EVALUATION ====================
print("\n" + "="*110)
print("RUNNING 10-FOLD CROSS-VALIDATION")
print("="*110)

kf = KFold(n_splits=10, shuffle=True, random_state=42)
results = []
cv_scores = {}  # Store all CV scores for statistical tests

print("\nEvaluating models (this may take a minute)...\n")

for name, model in models.items():
    print(f"{name:22s} → running...", end="")
    
    try:
        # Cross-validation scores
        accuracy_scores = cross_val_score(model, X_scaled, y, cv=kf, scoring='accuracy')
        precision_scores = cross_val_score(model, X_scaled, y, cv=kf, scoring='precision')
        recall_scores = cross_val_score(model, X_scaled, y, cv=kf, scoring='recall')
        f1_scores = cross_val_score(model, X_scaled, y, cv=kf, scoring='f1')
        roc_auc_scores = cross_val_score(model, X_scaled, y, cv=kf, scoring='roc_auc')
        
        # Store scores
        cv_scores[name] = {
            'accuracy': accuracy_scores,
            'precision': precision_scores,
            'recall': recall_scores,
            'f1': f1_scores,
            'roc_auc': roc_auc_scores
        }
        
        # Calculate means and stds
        results.append({
            'Model': name,
            'Accuracy': f"{accuracy_scores.mean():.4f} ± {accuracy_scores.std():.4f}",
            'Precision': f"{precision_scores.mean():.4f} ± {precision_scores.std():.4f}",
            'Recall': f"{recall_scores.mean():.4f} ± {recall_scores.std():.4f}",
            'F1-Score': f"{f1_scores.mean():.4f} ± {f1_scores.std():.4f}",
            'ROC-AUC': f"{roc_auc_scores.mean():.4f} ± {roc_auc_scores.std():.4f}",
            'Accuracy_Mean': accuracy_scores.mean(),
            'ROC_AUC_Mean': roc_auc_scores.mean()
        })
        
        print(" Done")
        
    except Exception as e:
        print(f" Error: {e}")

# ==================== 7. CREATE RESULTS DATAFRAME ====================
results_df = pd.DataFrame(results)
results_df = results_df.sort_values(by='Accuracy_Mean', ascending=False).reset_index(drop=True)
results_df['Rank'] = range(1, len(results_df) + 1)

# Reorder columns
display_cols = ['Rank', 'Model', 'Accuracy', 'Precision', 'Recall', 'F1-Score', 'ROC-AUC']
results_display = results_df[display_cols]

print("\n" + "="*110)
print("CLASSIFICATION RESULTS (10-FOLD CROSS-VALIDATION)")
print("="*110)

if IN_NOTEBOOK:
    from IPython.display import display
    display(results_display.style.background_gradient(cmap='viridis', subset=['Rank']))
else:
    print("\n" + results_display.to_string(index=False))

# Create output directory
output_dir = "breast_cancer_classification"
os.makedirs(output_dir, exist_ok=True)
timestamp = datetime.now().strftime("%Y%m%d_%H%M%S")

# Save results
results_df.to_excel(f"{output_dir}/Classification_Results_{timestamp}.xlsx", index=False)
results_df.to_csv(f"{output_dir}/Classification_Results_{timestamp}.csv", index=False)

# ==================== 8. STATISTICAL ANALYSIS ====================
print("\n" + "="*110)
print("STATISTICAL SIGNIFICANCE TESTING")
print("="*110)

# Friedman test
accuracy_scores_list = [cv_scores[name]['accuracy'] for name in cv_scores.keys()]
friedman_stat, friedman_p = friedmanchisquare(*accuracy_scores_list)

print(f"\nFriedman Test (Accuracy across all models):")
print(f"  Statistic: {friedman_stat:.4f}")
print(f"  P-value: {friedman_p:.4e}")
print(f"  Result: {'Significant differences' if friedman_p < 0.05 else 'No significant differences'} among models")

# Pairwise comparison (best model vs others)
best_model_name = results_df.iloc[0]['Model']
best_model_scores = cv_scores[best_model_name]['accuracy']

print(f"\nPairwise Wilcoxon Tests: Best model ({best_model_name}) vs Others")
print("-"*110)

pairwise_results = []
for name in cv_scores.keys():
    if name != best_model_name:
        try:
            stat, p_value = wilcoxon(best_model_scores, cv_scores[name]['accuracy'])
            pairwise_results.append({
                'Model': name,
                'vs': best_model_name,
                'P-value': f"{p_value:.4f}",
                'Significant': 'Yes' if p_value < 0.05 else 'No',
                'P_value_numeric': p_value
            })
        except:
            pass

pairwise_df = pd.DataFrame(pairwise_results)
pairwise_df = pairwise_df.sort_values('P_value_numeric').reset_index(drop=True)
print(pairwise_df[['Model', 'P-value', 'Significant']].to_string(index=False))

# ==================== 9. TRAIN BEST MODEL AND GET DETAILED METRICS ====================
print("\n" + "="*110)
print("DETAILED ANALYSIS - BEST MODEL")
print("="*110)

best_model = models[best_model_name]

# Split data for final evaluation
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42, stratify=y)

# Train best model
best_model.fit(X_train, y_train)
y_pred = best_model.predict(X_test)
y_pred_proba = best_model.predict_proba(X_test)[:, 1]

# Calculate metrics
test_accuracy = accuracy_score(y_test, y_pred)
test_precision = precision_score(y_test, y_pred)
test_recall = recall_score(y_test, y_pred)
test_f1 = f1_score(y_test, y_pred)
test_roc_auc = roc_auc_score(y_test, y_pred_proba)

print(f"\nBest Model: {best_model_name}")
print(f"\nTest Set Performance:")
print(f"  Accuracy:  {test_accuracy:.4f}")
print(f"  Precision: {test_precision:.4f}")
print(f"  Recall:    {test_recall:.4f}")
print(f"  F1-Score:  {test_f1:.4f}")
print(f"  ROC-AUC:   {test_roc_auc:.4f}")

# Confusion Matrix
cm = confusion_matrix(y_test, y_pred)
print(f"\nConfusion Matrix:")
print(f"  TN={cm[0,0]}, FP={cm[0,1]}")
print(f"  FN={cm[1,0]}, TP={cm[1,1]}")

# Classification Report
print(f"\nClassification Report:")
print(classification_report(y_test, y_pred, target_names=class_names))

# ==================== 10. FEATURE IMPORTANCE ====================
print("\n" + "="*110)
print("FEATURE IMPORTANCE ANALYSIS")
print("="*110)

has_feature_importance = False

# Check if model has feature_importances_ attribute (tree-based models)
if hasattr(best_model, 'feature_importances_'):
    try:
        importances = best_model.feature_importances_
        
        # Verify we have the right number of features
        if len(importances) != len(feature_names):
            print(f"\n⚠ Warning: Mismatch in feature importance length!")
            print(f"  Expected: {len(feature_names)} features")
            print(f"  Got: {len(importances)} importances")
            # Truncate or pad as needed
            if len(importances) > len(feature_names):
                importances = importances[:len(feature_names)]
            else:
                feature_names_temp = feature_names[:len(importances)]
        else:
            feature_names_temp = feature_names
        
        importance_df = pd.DataFrame({
            'Feature': feature_names_temp,
            'Importance': importances
        }).sort_values('Importance', ascending=False)
        
        print(f"\n✓ Feature importance extracted using feature_importances_")
        print(f"  Method: Tree-based importance (Gini/gain)")
        print(f"\nTop 10 Most Important Features:")
        print(importance_df.head(10).to_string(index=False))
        
        has_feature_importance = True
        
    except Exception as e:
        print(f"\n⚠ Error extracting feature importance: {e}")
        has_feature_importance = False

# Check if model has coef_ attribute (linear models)
elif hasattr(best_model, 'coef_') and best_model.coef_ is not None:
    try:
        coef = best_model.coef_
        
        # Handle multi-dimensional coefficients
        if len(coef.shape) > 1:
            if coef.shape[0] == 1:
                # Binary classification with single row
                coef = coef[0]
            else:
                # Multi-class: use first class or average
                print(f"\n  Note: Multi-class coefficients detected (shape: {coef.shape})")
                coef = coef[0]  # Use first class
        
        # Verify length matches features
        if len(coef) != len(feature_names):
            print(f"\n⚠ Warning: Coefficient length mismatch!")
            print(f"  Expected: {len(feature_names)} features")
            print(f"  Got: {len(coef)} coefficients")
            # Adjust feature names if needed
            feature_names_temp = feature_names[:len(coef)]
        else:
            feature_names_temp = feature_names
        
        # Use absolute values as importance
        importances = np.abs(coef)
        
        importance_df = pd.DataFrame({
            'Feature': feature_names_temp,
            'Importance': importances
        }).sort_values('Importance', ascending=False)
        
        print(f"\n✓ Feature importance extracted using coefficients (coef_)")
        print(f"  Method: Absolute coefficient values |w|")
        print(f"  Note: Linear model - coefficients represent feature weights")
        print(f"\nTop 10 Most Important Features (by |coefficient|):")
        print(importance_df.head(10).to_string(index=False))
        
        has_feature_importance = True
        
    except Exception as e:
        print(f"\n⚠ Error extracting coefficients: {e}")
        print(f"  Coefficient shape: {best_model.coef_.shape if hasattr(best_model, 'coef_') else 'N/A'}")
        has_feature_importance = False

# No direct feature importance available
else:
    print(f"\n⚠ {best_model_name} does not provide direct feature importance")
    print(f"  Model type: {type(best_model).__name__}")
    print(f"  Kernel: {best_model.kernel if hasattr(best_model, 'kernel') else 'N/A'}")
    print(f"\n  Models with direct feature importance:")
    print(f"    ✓ Tree-based: XGBoost, LightGBM, CatBoost, Random Forest, Extra Trees, Gradient Boosting")
    print(f"    ✓ Linear: Logistic Regression, SVM Linear")
    print(f"    ✗ Non-linear: SVM RBF, MLP Neural Network")
    print(f"\n  Note: For non-linear models, consider using:")
    print(f"    - Permutation importance (model-agnostic)")
    print(f"    - SHAP values (explains individual predictions)")
    print(f"    - Partial dependence plots")

# ==================== 11. NATURE MEDICINE STANDARD VISUALIZATIONS (MODEL RESULTS) ====================
print("\n" + "="*110)
print("CREATING MODEL PERFORMANCE VISUALIZATIONS")
print("="*110)

# Figure 6: Model Performance Comparison
fig, axes = plt.subplots(1, 2, figsize=(DOUBLE_COLUMN_WIDTH, DOUBLE_COLUMN_WIDTH * 0.35))

# Extract scores
model_names_sorted = results_df['Model'].values
accuracy_means = [cv_scores[name]['accuracy'].mean() for name in model_names_sorted]
roc_auc_means = [cv_scores[name]['roc_auc'].mean() for name in model_names_sorted]

y_pos = np.arange(len(model_names_sorted))

# Accuracy
axes[0].barh(y_pos, accuracy_means, color=NATURE_COLORS[0], alpha=0.85, edgecolor='black', linewidth=0.5)
axes[0].set_yticks(y_pos)
axes[0].set_yticklabels(model_names_sorted, fontsize=6)
axes[0].set_xlabel('Accuracy', fontsize=7)
axes[0].set_title('Model Accuracy (10-Fold CV)', fontsize=8, fontweight='bold')
axes[0].set_xlim([0.8, 1.0])
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)
axes[0].grid(axis='x', linestyle='--', alpha=0.3, linewidth=0.5)

# ROC-AUC
axes[1].barh(y_pos, roc_auc_means, color=NATURE_COLORS[1], alpha=0.85, edgecolor='black', linewidth=0.5)
axes[1].set_yticks(y_pos)
axes[1].set_yticklabels(model_names_sorted, fontsize=6)
axes[1].set_xlabel('ROC-AUC', fontsize=7)
axes[1].set_title('Model ROC-AUC (10-Fold CV)', fontsize=8, fontweight='bold')
axes[1].set_xlim([0.8, 1.0])
axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)
axes[1].grid(axis='x', linestyle='--', alpha=0.3, linewidth=0.5)

plt.tight_layout(pad=0.5)
plt.savefig(f"{output_dir}/Fig6_Model_Comparison_{timestamp}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{output_dir}/Fig6_Model_Comparison_{timestamp}.eps", format='eps', bbox_inches='tight')
if IN_NOTEBOOK:
    plt.show()
else:
    plt.close()
print("✓ Fig6: Model comparison saved")

# Figure 7: Confusion Matrix
fig, ax = plt.subplots(figsize=(DOUBLE_COLUMN_WIDTH * 0.4, DOUBLE_COLUMN_WIDTH * 0.35))

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', 
            xticklabels=class_names, yticklabels=class_names,
            ax=ax, cbar_kws={'label': 'Count', 'shrink': 0.8},
            linewidths=0.5, linecolor='white', annot_kws={'fontsize': 8})

ax.set_xlabel('Predicted Label', fontsize=7)
ax.set_ylabel('True Label', fontsize=7)
ax.set_title(f'Confusion Matrix - {best_model_name}', fontsize=8, fontweight='bold')
plt.setp(ax.get_xticklabels(), fontsize=6)
plt.setp(ax.get_yticklabels(), fontsize=6, rotation=0)

plt.tight_layout()
plt.savefig(f"{output_dir}/Fig7_Confusion_Matrix_{timestamp}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{output_dir}/Fig7_Confusion_Matrix_{timestamp}.eps", format='eps', bbox_inches='tight')
if IN_NOTEBOOK:
    plt.show()
else:
    plt.close()
print("✓ Fig7: Confusion matrix saved")

# Figure 8: ROC Curves for ALL Models
print("\n[Generating ROC curves for all models - this may take a moment...]")

fig, ax = plt.subplots(figsize=(DOUBLE_COLUMN_WIDTH * 0.5, DOUBLE_COLUMN_WIDTH * 0.45))

# Store all model ROC data
all_roc_data = []

for idx, (name, model) in enumerate(models.items()):
    try:
        # Train on full training set
        model_temp = model
        model_temp.fit(X_train, y_train)
        y_proba_temp = model_temp.predict_proba(X_test)[:, 1]
        
        fpr_temp, tpr_temp, _ = roc_curve(y_test, y_proba_temp)
        auc_temp = roc_auc_score(y_test, y_proba_temp)
        
        # Plot ROC curve
        ax.plot(fpr_temp, tpr_temp, 
                color=NATURE_COLORS[idx % len(NATURE_COLORS)], 
                linewidth=1.0, 
                alpha=0.8,
                label=f'{name} (AUC={auc_temp:.3f})')
        
        all_roc_data.append({
            'Model': name,
            'FPR': fpr_temp,
            'TPR': tpr_temp,
            'AUC': auc_temp
        })
    except Exception as e:
        print(f"  Warning: Could not plot ROC for {name}: {e}")

# Plot random classifier
ax.plot([0, 1], [0, 1], 'k--', linewidth=0.5, alpha=0.5, label='Random (AUC=0.5)')

ax.set_xlabel('False Positive Rate', fontsize=7)
ax.set_ylabel('True Positive Rate', fontsize=7)
ax.set_title('ROC Curves - All Models Comparison', fontsize=8, fontweight='bold')
ax.legend(fontsize=5, loc='lower right', framealpha=0.9)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(True, linestyle='--', alpha=0.3, linewidth=0.5)
ax.set_xlim([-0.01, 1.01])
ax.set_ylim([-0.01, 1.01])

plt.tight_layout()
plt.savefig(f"{output_dir}/Fig8_ROC_Curves_All_Models_{timestamp}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{output_dir}/Fig8_ROC_Curves_All_Models_{timestamp}.eps", format='eps', bbox_inches='tight')
if IN_NOTEBOOK:
    plt.show()
else:
    plt.close()
print("✓ Fig8: ROC curves (all models) saved")

# Figure 9: Feature Importance (if available)
if has_feature_importance:
    print("\n[Generating feature importance plot...]")
    fig, ax = plt.subplots(figsize=(DOUBLE_COLUMN_WIDTH * 0.5, DOUBLE_COLUMN_WIDTH * 0.5))
    
    top_n = min(15, len(importance_df))
    top_features = importance_df.head(top_n).sort_values('Importance', ascending=True)
    
    ax.barh(top_features['Feature'], top_features['Importance'], 
            color=NATURE_COLORS[3], alpha=0.85, edgecolor='black', linewidth=0.5)
    ax.set_xlabel('Importance Score', fontsize=7)
    ax.set_title(f'Top {top_n} Important Features - {best_model_name}', fontsize=8, fontweight='bold')
    ax.tick_params(axis='y', labelsize=6)
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.grid(axis='x', linestyle='--', alpha=0.3, linewidth=0.5)
    
    plt.tight_layout()
    plt.savefig(f"{output_dir}/Fig9_Feature_Importance_{timestamp}.png", dpi=300, bbox_inches='tight')
    plt.savefig(f"{output_dir}/Fig9_Feature_Importance_{timestamp}.eps", format='eps', bbox_inches='tight')
    if IN_NOTEBOOK:
        plt.show()
    else:
        plt.close()
    print("✓ Fig9: Feature importance saved")
else:
    print("\n⚠ Skipping Fig9: Best model does not support direct feature importance extraction")
    print(f"  Best Model: {best_model_name}")
    print(f"  Model Type: {type(best_model).__name__}")
    if hasattr(best_model, 'kernel'):
        print(f"  Kernel: {best_model.kernel}")
    print(f"\n  ✓ Models WITH feature importance support:")
    print(f"     - Tree-based: XGBoost, LightGBM, CatBoost, Random Forest, Extra Trees, Gradient Boosting")
    print(f"     - Linear: Logistic Regression (L1/L2), SVM Linear")
    print(f"  ✗ Models WITHOUT direct importance:")
    print(f"     - Non-linear: SVM RBF, MLP Neural Network")
    print(f"\n  Alternative methods for {best_model_name}:")
    print(f"     - Use permutation importance (sklearn.inspection.permutation_importance)")
    print(f"     - Use SHAP values for model interpretation")
    print(f"     - Re-run with a tree-based model as best model")

# Figure 10: Precision, Recall, F1 Comparison for All Models
fig, axes = plt.subplots(1, 3, figsize=(DOUBLE_COLUMN_WIDTH, DOUBLE_COLUMN_WIDTH * 0.35))

precision_means = [cv_scores[name]['precision'].mean() for name in model_names_sorted]
recall_means = [cv_scores[name]['recall'].mean() for name in model_names_sorted]
f1_means = [cv_scores[name]['f1'].mean() for name in model_names_sorted]

y_pos = np.arange(len(model_names_sorted))

# Precision
axes[0].barh(y_pos, precision_means, color=NATURE_COLORS[4], alpha=0.85, edgecolor='black', linewidth=0.5)
axes[0].set_yticks(y_pos)
axes[0].set_yticklabels(model_names_sorted, fontsize=6)
axes[0].set_xlabel('Precision', fontsize=7)
axes[0].set_title('Precision (10-Fold CV)', fontsize=8, fontweight='bold')
axes[0].set_xlim([0.85, 1.0])
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)
axes[0].grid(axis='x', linestyle='--', alpha=0.3, linewidth=0.5)

# Recall
axes[1].barh(y_pos, recall_means, color=NATURE_COLORS[5], alpha=0.85, edgecolor='black', linewidth=0.5)
axes[1].set_yticks(y_pos)
axes[1].set_yticklabels(model_names_sorted, fontsize=6)
axes[1].set_xlabel('Recall', fontsize=7)
axes[1].set_title('Recall (10-Fold CV)', fontsize=8, fontweight='bold')
axes[1].set_xlim([0.85, 1.0])
axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)
axes[1].grid(axis='x', linestyle='--', alpha=0.3, linewidth=0.5)

# F1-Score
axes[2].barh(y_pos, f1_means, color=NATURE_COLORS[6], alpha=0.85, edgecolor='black', linewidth=0.5)
axes[2].set_yticks(y_pos)
axes[2].set_yticklabels(model_names_sorted, fontsize=6)
axes[2].set_xlabel('F1-Score', fontsize=7)
axes[2].set_title('F1-Score (10-Fold CV)', fontsize=8, fontweight='bold')
axes[2].set_xlim([0.85, 1.0])
axes[2].spines['top'].set_visible(False)
axes[2].spines['right'].set_visible(False)
axes[2].grid(axis='x', linestyle='--', alpha=0.3, linewidth=0.5)

plt.tight_layout(pad=0.5)
plt.savefig(f"{output_dir}/Fig10_Precision_Recall_F1_{timestamp}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{output_dir}/Fig10_Precision_Recall_F1_{timestamp}.eps", format='eps', bbox_inches='tight')
if IN_NOTEBOOK:
    plt.show()
else:
    plt.close()
print("✓ Fig10: Precision, Recall, F1 comparison saved")

# Figure 11: Model Performance Distribution (Boxplot)
fig, axes = plt.subplots(1, 2, figsize=(DOUBLE_COLUMN_WIDTH, DOUBLE_COLUMN_WIDTH * 0.35))

# Prepare data for boxplot
accuracy_data = [cv_scores[name]['accuracy'] for name in model_names_sorted]
roc_auc_data = [cv_scores[name]['roc_auc'] for name in model_names_sorted]

# Accuracy boxplot
bp1 = axes[0].boxplot(accuracy_data, labels=model_names_sorted, patch_artist=True,
                       widths=0.6, showfliers=False,
                       boxprops=dict(linewidth=0.5),
                       whiskerprops=dict(linewidth=0.5),
                       capprops=dict(linewidth=0.5),
                       medianprops=dict(linewidth=1.0, color='red'))

for patch, color in zip(bp1['boxes'], NATURE_COLORS[:len(model_names_sorted)]):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)
    patch.set_edgecolor('black')

axes[0].set_ylabel('Accuracy', fontsize=7)
axes[0].set_title('Accuracy Distribution (10-Fold CV)', fontsize=8, fontweight='bold')
axes[0].set_ylim([0.85, 1.0])
axes[0].tick_params(axis='x', rotation=45, labelsize=6)
axes[0].tick_params(axis='y', labelsize=6)
axes[0].spines['top'].set_visible(False)
axes[0].spines['right'].set_visible(False)
axes[0].grid(axis='y', linestyle='--', alpha=0.3, linewidth=0.5)

# ROC-AUC boxplot
bp2 = axes[1].boxplot(roc_auc_data, labels=model_names_sorted, patch_artist=True,
                       widths=0.6, showfliers=False,
                       boxprops=dict(linewidth=0.5),
                       whiskerprops=dict(linewidth=0.5),
                       capprops=dict(linewidth=0.5),
                       medianprops=dict(linewidth=1.0, color='red'))

for patch, color in zip(bp2['boxes'], NATURE_COLORS[:len(model_names_sorted)]):
    patch.set_facecolor(color)
    patch.set_alpha(0.7)
    patch.set_edgecolor('black')

axes[1].set_ylabel('ROC-AUC', fontsize=7)
axes[1].set_title('ROC-AUC Distribution (10-Fold CV)', fontsize=8, fontweight='bold')
axes[1].set_ylim([0.85, 1.0])
axes[1].tick_params(axis='x', rotation=45, labelsize=6)
axes[1].tick_params(axis='y', labelsize=6)
axes[1].spines['top'].set_visible(False)
axes[1].spines['right'].set_visible(False)
axes[1].grid(axis='y', linestyle='--', alpha=0.3, linewidth=0.5)

plt.tight_layout(pad=0.5)
plt.savefig(f"{output_dir}/Fig11_Performance_Distribution_{timestamp}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{output_dir}/Fig11_Performance_Distribution_{timestamp}.eps", format='eps', bbox_inches='tight')
if IN_NOTEBOOK:
    plt.show()
else:
    plt.close()
print("✓ Fig11: Performance distribution saved")

# Figure 12: All Metrics Comparison Heatmap
fig, ax = plt.subplots(figsize=(DOUBLE_COLUMN_WIDTH * 0.6, DOUBLE_COLUMN_WIDTH * 0.5))

# Create metrics matrix
metrics_matrix = np.array([
    accuracy_means,
    precision_means,
    recall_means,
    f1_means,
    roc_auc_means
])

metric_labels = ['Accuracy', 'Precision', 'Recall', 'F1-Score', 'ROC-AUC']

sns.heatmap(metrics_matrix, annot=True, fmt='.3f', cmap='viridis',
            xticklabels=model_names_sorted, yticklabels=metric_labels,
            ax=ax, cbar_kws={'label': 'Score', 'shrink': 0.8},
            linewidths=0.5, linecolor='white', annot_kws={'fontsize': 5},
            vmin=0.9, vmax=1.0)

ax.set_title('Performance Metrics Heatmap - All Models', fontsize=8, fontweight='bold', pad=10)
plt.setp(ax.get_xticklabels(), rotation=45, ha='right', fontsize=6)
plt.setp(ax.get_yticklabels(), rotation=0, fontsize=6)

plt.tight_layout()
plt.savefig(f"{output_dir}/Fig12_Metrics_Heatmap_{timestamp}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{output_dir}/Fig12_Metrics_Heatmap_{timestamp}.eps", format='eps', bbox_inches='tight')
if IN_NOTEBOOK:
    plt.show()
else:
    plt.close()
print("✓ Fig12: Metrics heatmap saved")

# Figure 13: Model Comparison Boxplot with Statistical Significance (P-values)
print("\n[Creating statistical comparison boxplot with p-values...]")

fig, ax = plt.subplots(figsize=(DOUBLE_COLUMN_WIDTH, DOUBLE_COLUMN_WIDTH * 0.55))

# Prepare data for boxplot
accuracy_data = [cv_scores[name]['accuracy'] for name in model_names_sorted]

# Create boxplot
bp = ax.boxplot(accuracy_data, labels=model_names_sorted, patch_artist=True,
                widths=0.5, showfliers=True,
                boxprops=dict(linewidth=0.5),
                whiskerprops=dict(linewidth=0.5),
                capprops=dict(linewidth=0.5),
                medianprops=dict(linewidth=1.0, color='darkred'),
                flierprops=dict(marker='o', markerfacecolor='gray', markersize=2, alpha=0.5))

# Color the boxes
for idx, (patch, color) in enumerate(zip(bp['boxes'], NATURE_COLORS[:len(model_names_sorted)])):
    patch.set_facecolor(color)
    patch.set_alpha(0.75)
    patch.set_edgecolor('black')

# Calculate proper y-axis limits and offset for p-values
y_min = min([min(data) for data in accuracy_data])
y_max = max([max(data) for data in accuracy_data])
y_range = y_max - y_min
y_plot_max = y_max + y_range * 0.25  # Add 25% space for annotations

# Set y-axis limits
ax.set_ylim([max(0.85, y_min - y_range * 0.05), y_plot_max])

# Add statistical significance annotations (comparing each model to the best model)
best_model_accuracy = cv_scores[model_names_sorted[0]]['accuracy']
y_text_start = y_max + y_range * 0.02

# Add significance markers above each box (except the best model)
for idx, name in enumerate(model_names_sorted[1:], start=1):
    try:
        stat, p_val = wilcoxon(best_model_accuracy, cv_scores[name]['accuracy'])
        
        # Determine significance level and color
        if p_val < 0.001:
            sig_marker = '***'
            text_color = 'darkred'
        elif p_val < 0.01:
            sig_marker = '**'
            text_color = 'red'
        elif p_val < 0.05:
            sig_marker = '*'
            text_color = 'orange'
        else:
            sig_marker = 'ns'
            text_color = 'gray'
        
        # Calculate y position with staggering to avoid overlap
        y_pos = y_text_start + (idx % 3) * y_range * 0.04
        
        # Add p-value text above the box
        ax.text(idx + 1, y_pos, f'{sig_marker}\n(p={p_val:.3f})', 
                ha='center', va='bottom', fontsize=4.5, color=text_color,
                bbox=dict(boxstyle='round,pad=0.2', facecolor='white', 
                         edgecolor=text_color, alpha=0.85, linewidth=0.5))
        
    except Exception as e:
        print(f"  Warning: Could not compute p-value for {name}")

# Add best model label
ax.text(1, y_text_start, 'BEST', ha='center', va='bottom', fontsize=5,
        fontweight='bold', color='darkgreen',
        bbox=dict(boxstyle='round,pad=0.3', facecolor='gold', 
                 edgecolor='darkgreen', alpha=0.9, linewidth=0.8))

ax.set_ylabel('Accuracy (10-Fold CV)', fontsize=7, fontweight='bold')
ax.set_xlabel('Classification Models', fontsize=7)
ax.set_title('Model Comparison with Statistical Significance\n(Wilcoxon Test vs Best Model)', 
             fontsize=8, fontweight='bold', pad=10)
ax.tick_params(axis='x', rotation=45, labelsize=5.5)
ax.tick_params(axis='y', labelsize=6)
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.grid(axis='y', linestyle='--', alpha=0.3, linewidth=0.5)

# Add comprehensive legend at bottom
legend_text = ('Statistical Significance:  *** p<0.001 (highly significant)  |  ** p<0.01 (very significant)  |  '
               '* p<0.05 (significant)  |  ns: not significant (p≥0.05)')
fig.text(0.5, 0.01, legend_text, ha='center', fontsize=5.5, 
         bbox=dict(boxstyle='round,pad=0.4', facecolor='lightyellow', 
                  edgecolor='black', alpha=0.95, linewidth=0.5))

plt.tight_layout(rect=[0, 0.05, 1, 1])  # Make room for legend at bottom
plt.savefig(f"{output_dir}/Fig13_Statistical_Comparison_{timestamp}.png", dpi=300, bbox_inches='tight')
plt.savefig(f"{output_dir}/Fig13_Statistical_Comparison_{timestamp}.eps", format='eps', bbox_inches='tight')
if IN_NOTEBOOK:
    plt.show()
else:
    plt.close()
print("✓ Fig13: Statistical comparison boxplot with p-values saved")

# ==================== 12. SAVE COMPREHENSIVE EXCEL REPORT (ALL MODELS) ====================
print("\n" + "="*110)
print("SAVING COMPREHENSIVE REPORT WITH ALL MODELS DATA")
print("="*110)

with pd.ExcelWriter(f"{output_dir}/Comprehensive_Classification_Report_{timestamp}.xlsx", engine='openpyxl') as writer:
    # Sheet 1: Model Performance Summary
    results_df.to_excel(writer, sheet_name='Model_Performance', index=False)
    
    # Sheet 2: Detailed Metrics (All Models)
    detailed_metrics_data = []
    for name in model_names_sorted:
        detailed_metrics_data.append({
            'Model': name,
            'Accuracy_Mean': cv_scores[name]['accuracy'].mean(),
            'Accuracy_Std': cv_scores[name]['accuracy'].std(),
            'Precision_Mean': cv_scores[name]['precision'].mean(),
            'Precision_Std': cv_scores[name]['precision'].std(),
            'Recall_Mean': cv_scores[name]['recall'].mean(),
            'Recall_Std': cv_scores[name]['recall'].std(),
            'F1_Mean': cv_scores[name]['f1'].mean(),
            'F1_Std': cv_scores[name]['f1'].std(),
            'ROC_AUC_Mean': cv_scores[name]['roc_auc'].mean(),
            'ROC_AUC_Std': cv_scores[name]['roc_auc'].std(),
        })
    
    detailed_metrics_df = pd.DataFrame(detailed_metrics_data)
    detailed_metrics_df.to_excel(writer, sheet_name='Detailed_Metrics_All_Models', index=False)
    
    # Sheet 3: Per-Fold Results (All Models)
    per_fold_data = []
    for name in model_names_sorted:
        for fold_idx in range(10):
            per_fold_data.append({
                'Model': name,
                'Fold': fold_idx + 1,
                'Accuracy': cv_scores[name]['accuracy'][fold_idx],
                'Precision': cv_scores[name]['precision'][fold_idx],
                'Recall': cv_scores[name]['recall'][fold_idx],
                'F1_Score': cv_scores[name]['f1'][fold_idx],
                'ROC_AUC': cv_scores[name]['roc_auc'][fold_idx]
            })
    
    per_fold_df = pd.DataFrame(per_fold_data)
    per_fold_df.to_excel(writer, sheet_name='Per_Fold_Results', index=False)
    
    # Sheet 4: Statistical Tests
    pairwise_df[['Model', 'P-value', 'Significant']].to_excel(writer, sheet_name='Statistical_Tests', index=False)
    
    # Sheet 5: Best Model Details
    best_model_details = pd.DataFrame({
        'Metric': ['Model Name', 'Test Accuracy', 'Test Precision', 'Test Recall', 'Test F1-Score', 'Test ROC-AUC',
                   'True Negatives', 'False Positives', 'False Negatives', 'True Positives',
                   'Sensitivity (TPR)', 'Specificity (TNR)', 'False Positive Rate', 'False Negative Rate'],
        'Value': [best_model_name, test_accuracy, test_precision, test_recall, test_f1, test_roc_auc,
                  cm[0,0], cm[0,1], cm[1,0], cm[1,1],
                  cm[1,1]/(cm[1,1]+cm[1,0]) if (cm[1,1]+cm[1,0])>0 else 0,
                  cm[0,0]/(cm[0,0]+cm[0,1]) if (cm[0,0]+cm[0,1])>0 else 0,
                  cm[0,1]/(cm[0,0]+cm[0,1]) if (cm[0,0]+cm[0,1])>0 else 0,
                  cm[1,0]/(cm[1,1]+cm[1,0]) if (cm[1,1]+cm[1,0])>0 else 0]
    })
    best_model_details.to_excel(writer, sheet_name='Best_Model_Details', index=False)
    
    # Sheet 6: Feature Importance
    if has_feature_importance:
        importance_df.to_excel(writer, sheet_name='Feature_Importance', index=False)
    
    # Sheet 7: ROC Data (All Models)
    if len(all_roc_data) > 0:
        roc_summary = pd.DataFrame([
            {'Model': item['Model'], 'AUC': item['AUC']} 
            for item in all_roc_data
        ]).sort_values('AUC', ascending=False)
        roc_summary.to_excel(writer, sheet_name='ROC_AUC_All_Models', index=False)
    
    # Sheet 8: Summary
    summary = pd.DataFrame({
        'Item': ['Dataset', 'Total Samples', 'Number of Features', 'Class 0 (Benign)', 'Class 1 (Malignant)',
                 'Number of Models Evaluated', 'Best Model', 'Best Accuracy', 'Best ROC-AUC', 'Friedman Test P-value'],
        'Value': ['SKLEARN BUILT-IN', X.shape[0], X.shape[1], np.sum(y==0), np.sum(y==1),
                  len(models), best_model_name, f"{accuracy_means[0]:.4f}", f"{roc_auc_means[0]:.4f}", f"{friedman_p:.4e}"]
    })
    summary.to_excel(writer, sheet_name='Summary', index=False)

print(f"✓ Comprehensive Excel report with {len(models)} models saved (8 sheets)")

print(f"\n{'='*110}")
print("ANALYSIS COMPLETE!")
print(f"{'='*110}")
print(f"\n✓ All results saved to: {output_dir}/")
print(f"\n✓ Excel report: Comprehensive_Classification_Report_{timestamp}.xlsx (8 sheets)")
print(f"   - Sheet 1: Model Performance Summary (all {len(models)} models)")
print(f"   - Sheet 2: Detailed Metrics with Mean±Std for all models")
print(f"   - Sheet 3: Per-Fold Results (10 folds × {len(models)} models = {10*len(models)} rows)")
print(f"   - Sheet 4: Statistical Tests (Friedman & Wilcoxon)")
print(f"   - Sheet 5: Best Model Details (with Sensitivity/Specificity)")
print(f"   - Sheet 6: Feature Importance")
print(f"   - Sheet 7: ROC-AUC Data for all {len(models)} models")
print(f"   - Sheet 8: Summary")
print(f"\n✓ Figures (PNG + EPS): 13 publication-ready figures")
print(f"   DATA EXPLORATION (5 figures):")
print(f"   - DataViz_01: Class Distribution (count + pie chart)")
print(f"   - DataViz_02: Feature Statistics (mean ± std)")
print(f"   - DataViz_03: Feature Distribution by Class (top 6 features)")
print(f"   - DataViz_04: Correlation Heatmap (first 15 features)")
print(f"   - DataViz_05: Data Quality Summary")
print(f"   MODEL PERFORMANCE (8 figures):")
print(f"   - Fig6: Model Accuracy & ROC-AUC Comparison")
print(f"   - Fig7: Confusion Matrix (Best Model)")
print(f"   - Fig8: ROC Curves for ALL {len(models)} Models")
print(f"   - Fig9: Feature Importance")
print(f"   - Fig10: Precision/Recall/F1 Comparison (All Models)")
print(f"   - Fig11: Performance Distribution Boxplots")
print(f"   - Fig12: Metrics Heatmap (5 metrics × {len(models)} models)")
print(f"   - Fig13: Statistical Comparison Boxplot with P-values (Wilcoxon Tests)")
print(f"\n📊 KEY FINDINGS:")
print(f"  • Dataset: SKLEARN BUILT-IN ({X.shape[0]} samples, {X.shape[1]} features)")
print(f"  • Models Evaluated: {len(models)}")
print(f"  • Best Model: {best_model_name}")
print(f"  • Best Cross-Val Accuracy: {accuracy_means[0]:.4f} ± {cv_scores[model_names_sorted[0]]['accuracy'].std():.4f}")
print(f"  • Best Cross-Val ROC-AUC: {roc_auc_means[0]:.4f} ± {cv_scores[model_names_sorted[0]]['roc_auc'].std():.4f}")
print(f"  • Test Set Accuracy: {test_accuracy:.4f}")
print(f"  • Test Set ROC-AUC: {test_roc_auc:.4f}")
print(f"  • Significant differences among models: {'Yes' if friedman_p < 0.05 else 'No'} (p={friedman_p:.4e})")
if has_feature_importance:
    print(f"  • Top 3 Features: {', '.join(importance_df.head(3)['Feature'].values)}")
print(f"\n✅ Ready for publication!")
print(f"✅ All figures follow Nature Medicine standards (300 DPI, Arial font, EPS format)")
print(f"✅ Comprehensive Excel report includes detailed per-fold results for ALL models")

# ==================== 13. CREATE ZIP FILE FOR DOWNLOAD ====================
print(f"\n{'='*110}")
print("CREATING ZIP FILE FOR DOWNLOAD")
print(f"{'='*110}")

import shutil
from pathlib import Path

# Create ZIP file with all outputs
zip_filename = f"breast_cancer_analysis_results_{timestamp}"
zip_path = f"{zip_filename}.zip"

print(f"\nCompressing all outputs into ZIP file...")
print(f"  Source directory: {output_dir}/")
print(f"  ZIP file: {zip_path}")

try:
    # Create ZIP archive
    shutil.make_archive(zip_filename, 'zip', output_dir)
    
    # Get ZIP file size
    zip_size = os.path.getsize(zip_path) / (1024 * 1024)  # Convert to MB
    
    print(f"\n✓ ZIP file created successfully!")
    print(f"  File name: {zip_path}")
    print(f"  File size: {zip_size:.2f} MB")
    print(f"  Location: {os.path.abspath(zip_path)}")
    
    # Count files in output directory
    file_count = len([f for f in os.listdir(output_dir) if os.path.isfile(os.path.join(output_dir, f))])
    
    print(f"\n📦 ZIP Contents:")
    print(f"  Total files: {file_count}")
    print(f"  - Data visualizations: 5 figures (PNG + EPS)")
    print(f"  - Model performance: 8 figures (PNG + EPS)")
    print(f"  - Excel reports: 2 files")
    print(f"  - CSV results: 1 file")
    
    print(f"\n💾 DOWNLOAD INSTRUCTIONS (Kaggle):")
    print(f"  1. Look for the ZIP file in the output section on the right →")
    print(f"  2. Click on '{zip_path}' to download")
    print(f"  3. Or copy the file from: /kaggle/working/{zip_path}")
    
    # List all files in the ZIP
    print(f"\n📋 Detailed file list:")
    for filename in sorted(os.listdir(output_dir)):
        file_path = os.path.join(output_dir, filename)
        if os.path.isfile(file_path):
            file_size_kb = os.path.getsize(file_path) / 1024
            print(f"  - {filename} ({file_size_kb:.1f} KB)")
    
except Exception as e:
    print(f"\n⚠ Error creating ZIP file: {e}")
    print(f"  Files are still available in: {output_dir}/")

print(f"\n{'='*110}")
```
